# Why Train Multiple Machine Learning Models?

## Objective

There is no single machine learning algorithm that performs best for every dataset.

Different models make different assumptions about the data and learn patterns in different ways. Comparing multiple algorithms helps identify the model that offers the best balance between predictive performance, interpretability, and robustness.

In this notebook, the following classification models will be trained:

- **Logistic Regression** – A simple and interpretable baseline model.
- **Decision Tree** – Captures non-linear relationships through hierarchical decision rules.
- **Random Forest** – An ensemble of decision trees that improves robustness and reduces overfitting.

The trained models will be evaluated and compared in the next notebook to determine the most suitable model for employee attrition prediction.

## Models to be Trained

| Model | Purpose |
|--------|---------|
| Logistic Regression | Establish a simple and interpretable baseline model |
| Decision Tree | Learn non-linear decision boundaries |
| Random Forest | Improve predictive performance using ensemble learning |

# Model 1: Logistic Regression

## Objective

Logistic Regression is used as the baseline classification model for this project.

It estimates the probability that an employee will leave the organization by applying the sigmoid function to a linear combination of the input features.

Although it is a simple algorithm, Logistic Regression is fast, interpretable, and provides a strong benchmark for comparing more complex models.

```python
# Initialize Logistic Regression model

logistic_model = LogisticRegression(
    random_state=42,
    max_iter=1000
)
```

The model is initialized with a fixed random state to ensure reproducible results. The maximum number of iterations is increased to help the optimization algorithm converge successfully.

```python
# Train Logistic Regression model

logistic_model.fit(
    X_train_scaled,
    y_train
)
```

The Logistic Regression model has been trained using the scaled training dataset.

Feature scaling is applied because Logistic Regression is sensitive to differences in feature magnitudes.

```python
# Generate predictions

y_pred_lr = logistic_model.predict(
    X_test_scaled
)
```

```python
# Generate prediction probabilities

y_prob_lr = logistic_model.predict_proba(
    X_test_scaled
)[:, 1]
```

Prediction probabilities indicate the likelihood that an employee belongs to the **Attrition = Yes** class.

These probabilities will later be used to evaluate the model using metrics such as ROC-AUC and to visualize model performance.

```python
print("First 10 Predictions:\n")
print(y_pred_lr[:10])
```

```python
print("First 10 Prediction Probabilities:\n")
print(y_prob_lr[:10])
```

```python
print("Model Classes:")

print(logistic_model.classes_)
```

### Initial Observations

- The Logistic Regression model was trained successfully.
- Predictions for the testing dataset were generated without errors.
- Prediction probabilities are available and will support threshold-based evaluation and ROC curve analysis in the next notebook.
- As the baseline model, Logistic Regression provides a simple and interpretable benchmark against which more complex models can be compared.

## Understanding Logistic Regression

**Strengths**

- Simple and computationally efficient.
- Produces interpretable coefficients.
- Estimates class probabilities.
- Works well when the relationship between features and the target is approximately linear.

**Limitations**

- Assumes a linear decision boundary.
- May not capture complex non-linear relationships.
- Performance can be affected by multicollinearity and outliers.

For these reasons, additional models such as Decision Trees and Random Forests will also be trained and compared.

# Model 2: Decision Tree

## Objective

Decision Trees classify data by recursively splitting it into smaller groups based on feature values.

Unlike Logistic Regression, Decision Trees can capture complex and non-linear relationships between employee characteristics and attrition without requiring feature scaling.

This makes them useful for identifying decision rules that influence employee turnover.

A Decision Tree builds a hierarchy of decision rules by selecting the features that best separate the target classes at each split.

For this baseline model, the default tree parameters are used. Model complexity can be controlled later through hyperparameter tuning if overfitting is observed.

```python
# Initialize Decision Tree model

decision_tree = DecisionTreeClassifier(
    random_state=42
)
```

```python
# Train Decision Tree model

decision_tree.fit(
    X_train,
    y_train
)
```

The Decision Tree model has been trained using the unscaled training dataset.

Unlike distance-based and linear models, Decision Trees are not affected by differences in feature magnitudes, so feature scaling is not required.

```python
# Generate predictions

y_pred_dt = decision_tree.predict(
    X_test
)
```

```python
# Generate prediction probabilities

y_prob_dt = decision_tree.predict_proba(
    X_test
)[:, 1]
```

```python
print("First 10 Predictions:\n")
print(y_pred_dt[:10])
```

```python
print("First 10 Prediction Probabilities:\n")
print(y_prob_dt[:10])
```

```python
print("Model Classes:")

print(decision_tree.classes_)
```

### Initial Observations

- The Decision Tree model was trained successfully.
- Predictions and prediction probabilities were generated for the testing dataset.
- The model can naturally capture non-linear relationships and feature interactions.
- However, unrestricted Decision Trees may overfit the training data, making hyperparameter tuning important in later stages.

## Understanding Decision Trees

**Strengths**

- Easy to understand and visualize.
- Captures complex, non-linear relationships.
- Does not require feature scaling.
- Handles both numerical and categorical information (after encoding).

**Limitations**

- Can easily overfit the training data.
- Small changes in data may produce different tree structures.
- A single tree generally has lower stability than ensemble methods.

To overcome these limitations, the next model—Random Forest—combines multiple Decision Trees to improve generalization and predictive performance.

## Tree Complexity

Decision Trees can continue splitting until every training sample is classified correctly, which may lead to overfitting.

Common hyperparameters used to control tree complexity include:

- **max_depth** – Limits the maximum depth of the tree.
- **min_samples_split** – Minimum number of samples required to split a node.
- **min_samples_leaf** – Minimum number of samples required at a leaf node.

For this baseline implementation, the default parameters are used. Hyperparameter tuning can be performed later after comparing the baseline models.

# Model 3: Random Forest

## Objective

Random Forest is an ensemble learning algorithm that combines multiple Decision Trees to improve predictive performance and reduce overfitting.

Instead of relying on a single tree, Random Forest builds many trees using different subsets of the training data and features. The final prediction is determined by combining the predictions from all trees through majority voting.

Random Forest addresses one of the main limitations of Decision Trees—overfitting.

By introducing randomness in both the training samples and the selected features, it produces a more stable and generalizable model while maintaining the ability to capture complex, non-linear relationships.

```python
# Initialize Random Forest model

random_forest = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)
```

```python
# Train Random Forest model

random_forest.fit(
    X_train,
    y_train
)
```

The Random Forest model has been trained using the unscaled training dataset.

Since Random Forest is a tree-based algorithm, feature scaling is not required.

```python
# Generate predictions

y_pred_rf = random_forest.predict(
    X_test
)
```

```python
# Generate prediction probabilities

y_prob_rf = random_forest.predict_proba(
    X_test
)[:, 1]
```

```python
print("First 10 Predictions:\n")
print(y_pred_rf[:10])
```

```python
print("First 10 Prediction Probabilities:\n")
print(y_prob_rf[:10])
```

```python
print("Model Classes:")

print(random_forest.classes_)
```

### Initial Observations

- The Random Forest model was trained successfully.
- Predictions and prediction probabilities were generated for the testing dataset.
- By combining multiple Decision Trees, Random Forest is generally more robust and less prone to overfitting than a single Decision Tree.
- The model is expected to provide strong predictive performance while maintaining good generalization.

## Understanding Random Forest

**Strengths**

- Reduces overfitting compared to a single Decision Tree.
- Captures complex and non-linear relationships.
- Handles high-dimensional data effectively.
- Provides estimates of feature importance.
- Does not require feature scaling.

**Limitations**

- Less interpretable than Logistic Regression or a single Decision Tree.
- Training time is higher because multiple trees are built.
- Large forests require more memory and computational resources.

Despite these limitations, Random Forest is widely used because it offers a strong balance between accuracy, robustness, and ease of use.

## Ensemble Learning

Random Forest is based on the concept of **ensemble learning**, where multiple models work together to produce a better prediction than any individual model.

Key concepts include:

- **Bootstrap Sampling:** Each tree is trained on a random sample of the training data.
- **Feature Randomness:** Each split considers only a random subset of features.
- **Majority Voting:** The final class prediction is determined by the prediction receiving the most votes from all trees.

These techniques improve model stability and reduce variance, leading to better generalization on unseen data.
