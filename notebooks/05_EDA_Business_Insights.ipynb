# Correlation Analysis

## Objective

Analyze the relationships among numerical variables to identify patterns, dependencies, and potential factors associated with employee attrition.

```python
plt.figure(figsize=(15,10))

sns.heatmap(
    df.select_dtypes(include=np.number).corr(),
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Numerical Features")

plt.show()
```

- Experience-related variables such as **TotalWorkingYears**, **YearsAtCompany**, **YearsInCurrentRole**, and **YearsWithCurrManager** show positive correlations, which is expected.
- Monthly income is positively correlated with experience and job level, suggesting that more experienced employees generally earn higher salaries.
- Overall, the dataset contains a balanced mix of independent variables suitable for predictive modeling.

# Crosstab Analysis

## Objective

Calculate attrition percentages across different employee groups to identify categories with relatively higher employee turnover.

## Business Question 20

### What percentage of employees leave from each department?

**Objective**

Compare attrition rates across departments rather than relying only on employee counts.

```python
department_attrition = (
    pd.crosstab(
        df["Department"],
        df["Attrition"],
        normalize="index"
    ) * 100
).round(2)

department_attrition
```

### Observation

- Department-wise percentages provide a fairer comparison than raw employee counts.
- Departments with higher attrition percentages may require focused employee retention strategies.

## Business Question 21

### Which job roles experience the highest attrition rate?

**Objective**

Compare attrition percentages across job roles.

```python
jobrole_attrition = (
    pd.crosstab(
        df["JobRole"],
        df["Attrition"],
        normalize="index"
    ) * 100
).round(2)

jobrole_attrition
```

### Observation

- Some job roles exhibit noticeably higher attrition rates than others.
- Attrition percentages provide a more accurate comparison than employee counts because job roles differ in workforce size.

## Business Question 22

### Does overtime influence attrition percentage?

**Objective**

Compare attrition percentages between employees who work overtime and those who do not.

```python
overtime_attrition = (
    pd.crosstab(
        df["OverTime"],
        df["Attrition"],
        normalize="index"
    ) * 100
).round(2)

overtime_attrition
```

### Observation

- Employees working overtime typically exhibit a substantially higher attrition rate.
- This finding reinforces the earlier visual analysis and highlights overtime as a key factor associated with employee turnover.

## Business Question 23

# Final Business Insights

The bivariate and multivariate analysis highlights several factors that are associated with employee attrition and provides valuable insights into workforce behavior.

## Workforce Demographics

- Employees in the **26–35 years** age group experience the highest attrition, indicating that early-career professionals are more likely to leave.
- Attrition differences between male and female employees are relatively small, suggesting that gender is not a major driver of employee turnover.
- Marital status shows some variation in attrition, with certain groups appearing more likely to leave than others.

## Job Characteristics

- The **Research & Development** department records the highest number of employees leaving; however, this should be interpreted alongside its larger workforce size.
- Attrition varies across job roles, indicating that some roles are more susceptible to employee turnover than others.
- Employees who work **overtime** exhibit considerably higher attrition, making overtime one of the strongest indicators of employee turnover.
- Employees who travel frequently for business also tend to experience higher attrition than those who travel rarely.
- Lower job levels account for a larger proportion of employee attrition compared to senior job levels.

## Compensation

- Employees with lower monthly incomes are more likely to leave the organization than those with higher incomes.
- Salary hikes alone do not show a strong relationship with employee attrition.
- Employees with higher stock option levels generally exhibit lower attrition, suggesting that long-term financial incentives may improve employee retention.

## Experience

- Employees with fewer total working years and shorter company tenure are more likely to leave.
- Longer-tenured and more experienced employees generally demonstrate greater organizational stability.
- Delayed promotions may contribute to employee turnover, indicating the importance of career growth opportunities.
- Employees with longer commuting distances show slightly higher attrition, although the relationship is less pronounced than factors such as overtime or compensation.

## Employee Satisfaction

- Lower job satisfaction is associated with higher employee attrition.
- Employees reporting poorer work-life balance are more likely to leave the organization.
- Lower environment satisfaction and relationship satisfaction are also associated with increased attrition, emphasizing the importance of a positive workplace culture.

## Overall Conclusion

The analysis suggests that employee attrition is influenced by a combination of demographic, organizational, financial, experience, and satisfaction-related factors. Among these, **overtime, monthly income, job satisfaction, work-life balance, experience, and company tenure** emerge as the most influential factors. These insights provide a strong foundation for developing predictive machine learning models and can help HR departments design targeted employee retention strategies.
