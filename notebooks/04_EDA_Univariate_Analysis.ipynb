# Summary of Key Insights

The univariate analysis provides an overview of the employee population and highlights the overall characteristics of the dataset.

## Employee Demographics

- Most employees are between **30 and 40 years** of age.
- The workforce consists of both male and female employees, with males representing a larger proportion.
- Employees come from diverse educational backgrounds, with a few education fields accounting for the majority of the workforce.

## Experience and Career Progression

- Most employees have a moderate level of overall work experience.
- The majority have spent only a few years at the company, in their current role, and with their current manager.
- Most employees received a promotion within the last few years, while a small number have remained without promotion for a considerably longer period.

## Salary and Compensation

- Monthly income is positively skewed, with a small number of employees earning significantly higher salaries.
- Most employees fall within the low-to-middle income range.
- Salary hikes are concentrated within a relatively narrow range, indicating consistent compensation policies.

## Organization and Workforce Distribution

- Research & Development is the largest department, followed by Sales.
- Employees are distributed across multiple job roles, with a few roles accounting for most of the workforce.
- Most employees belong to the lower job levels, while higher job levels have comparatively fewer employees.

## Employee Characteristics

- Most employees travel rarely for business.
- Approximately 28% of employees work overtime, while the majority do not.
- Most employees report a work-life balance rating of **3**, indicating generally satisfactory work-life balance.
- Employees typically attend a moderate number of training programs each year.

## Attrition

- The target variable is imbalanced, with significantly more employees staying than leaving the organization.
- This imbalance should be considered during predictive model development and evaluation.
