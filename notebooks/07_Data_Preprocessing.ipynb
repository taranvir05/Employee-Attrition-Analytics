# Evaluate Features for Machine Learning

## Objective

Before preparing the dataset for machine learning, each feature is evaluated to determine whether it provides useful predictive information.

Identifier columns are removed because they uniquely identify employees rather than describe their characteristics. Such columns do not help the model learn patterns related to employee attrition and may introduce unnecessary noise.

The cleaned dataset created in the previous notebooks has already removed constant and irrelevant columns. Therefore, only the remaining identifier column is removed in this step.

```python
# Remove identifier column
df = df.drop(columns=["EmployeeNumber"], errors="ignore")

# Verify the updated dataset
df.shape
```

### Observation

The **EmployeeNumber** column has been removed because it serves only as a unique identifier and does not contribute to predicting employee attrition.

The dataset now contains only meaningful employee attributes that can be used for machine learning.

# Separate Features and Target

## Objective

Machine learning requires separating the dataset into:

- **Input Features (X):** Variables used to predict employee attrition.
- **Target Variable (y):** The outcome that the model will learn to predict.

In this project, **Attrition** is selected as the target variable, while all remaining columns serve as input features.

```python
# Separate features and target

X = df.drop(columns=["Attrition"])
y = df["Attrition"]

print("Feature matrix shape:", X.shape)
print("Target vector shape:", y.shape)
```

```python
X.head()
```

```python
y.head()
```

### Observation

The dataset has been successfully divided into input features (**X**) and the target variable (**y**).

- **X** contains all employee attributes that will be used for prediction.
- **y** contains the employee attrition status that the machine learning models will learn to classify.

# Encode the Target Variable

## Objective

Machine learning algorithms require numerical target values rather than text labels.

The target variable **Attrition** currently contains two categories:

- **Yes** – Employee left the company
- **No** – Employee stayed with the company

Since this is a binary classification problem, label encoding is appropriate.

The categories will be encoded as:

- **Yes → 1**
- **No → 0**

```python
# Encode target variable

y = y.map({
    "No": 0,
    "Yes": 1
})

y.head()
```

```python
# Verify target encoding

y.value_counts()
```

### Observation

The target variable has been successfully converted into numerical values.

- **0** represents employees who stayed with the company.
- **1** represents employees who left the company.

This format is suitable for binary classification algorithms.

# Identify Numerical and Categorical Features

## Objective

Different preprocessing techniques are required for numerical and categorical variables.

- **Numerical features** will later be scaled when required.
- **Categorical features** must be encoded before they can be used by machine learning algorithms.

Separating these feature types helps apply the appropriate preprocessing methods efficiently.

```python
# Identify numerical and categorical columns

numerical_features = X.select_dtypes(include=["int64", "float64"]).columns.tolist()

categorical_features = X.select_dtypes(include=["object", "category"]).columns.tolist()

print("Number of Numerical Features:", len(numerical_features))
print("Number of Categorical Features:", len(categorical_features))
```

```python
print("Numerical Features:\n")
print(numerical_features)
```

```python
print("Categorical Features:\n")
print(categorical_features)
```

### Observation

The dataset has been divided into numerical and categorical feature groups.

- **Numerical features** represent measurable employee attributes such as age, income, experience, and distance from home.
- **Categorical features** represent employee classifications such as department, job role, business travel, and the engineered category-based features.

This separation ensures that the appropriate preprocessing techniques can be applied in the following steps.

# Encode Categorical Features

## Objective

Machine learning algorithms cannot directly process text-based categorical variables.

Therefore, categorical features are converted into numerical representations using **One-Hot Encoding**.

One-Hot Encoding creates a separate binary column for each category without introducing any artificial ordering between categories, making it suitable for nominal variables.

```python
# Display categorical features

categorical_features
```

The following categorical features will be encoded using One-Hot Encoding.

This includes both original categorical variables and the engineered categorical features created during feature engineering.

```python
# Apply One-Hot Encoding

X_encoded = pd.get_dummies(
    X,
    columns=categorical_features,
    drop_first=True,
    dtype=int
)
```

```python
# Display encoded dataset

X_encoded.head()
```

```python
print("Original Number of Features :", X.shape[1])
print("Encoded Number of Features  :", X_encoded.shape[1])
```

```python
# Check remaining data types

X_encoded.dtypes.value_counts()
```

```python
# Verify that no object columns remain

X_encoded.select_dtypes(include="object").columns
```

### Observation

All categorical variables have been successfully converted into numerical features using One-Hot Encoding.

The number of input features has increased because each category is represented by its own binary column.

No object data types remain in the feature matrix, confirming that the dataset is now suitable for machine learning algorithms.

# Verify Encoded Dataset

Inspect the encoded feature matrix to ensure that the preprocessing step has been completed successfully before splitting the data into training and testing sets.

```python
X_encoded.info()
```

# Split the Dataset into Training and Testing Sets

## Objective

Split the dataset into separate training and testing sets.

- The **training set** is used to train machine learning models.
- The **testing set** is used to evaluate model performance on unseen data.

An **80:20 split** is used, and **stratified sampling** ensures that the proportion of attrition classes remains consistent in both datasets.

```python
# Split dataset into training and testing sets

X_train, X_test, y_train, y_test = train_test_split(
    X_encoded,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
print("Training Features :", X_train.shape)
print("Testing Features  :", X_test.shape)

print("Training Labels   :", y_train.shape)
print("Testing Labels    :", y_test.shape)
```

### Observation

The dataset has been successfully divided into training and testing sets.

Approximately **80%** of the data will be used for model training, while the remaining **20%** will be reserved for evaluating model performance on unseen data.

# Scale Numerical Features

## Objective

Many machine learning algorithms perform better when numerical features are on a similar scale.

Examples include:

- Logistic Regression
- Support Vector Machine (SVM)
- K-Nearest Neighbors (KNN)

Tree-based models such as Decision Trees and Random Forests do not require feature scaling.

To support both types of algorithms, scaled and unscaled datasets will be maintained.

```python
# Initialize StandardScaler

scaler = StandardScaler()
```

```python
# Create copies for scaling

X_train_scaled = X_train.copy()
X_test_scaled = X_test.copy()
```

```python
# Identify numerical columns after encoding

numerical_columns = numerical_features
```

```python
# Fit scaler on training data and transform

X_train_scaled[numerical_columns] = scaler.fit_transform(
    X_train[numerical_columns]
)
```

```python
# Transform testing data

X_test_scaled[numerical_columns] = scaler.transform(
    X_test[numerical_columns]
)
```

```python
# Display scaled training data

X_train_scaled.head()
```

### Observation

Only the numerical features have been standardized using **StandardScaler**.

The scaler was fitted **only on the training data** and then applied to the testing data, preventing data leakage and ensuring a fair evaluation of machine learning models.

# Verify Target Distribution

## Objective

Verify that stratified sampling has preserved the class distribution of the target variable in both the training and testing datasets.

```python
print("Training Target Distribution")

print(y_train.value_counts(normalize=True) * 100)
```

```python
print("Testing Target Distribution")

print(y_test.value_counts(normalize=True) * 100)
```

### Observation

The class distribution remains nearly identical in both the training and testing datasets, confirming that stratified sampling has been applied successfully.
