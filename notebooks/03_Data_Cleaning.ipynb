# Data Cleaning

## Objective

Identify and fix data quality issues while preserving as much useful information as possible.

```python
import pandas as pd
import numpy as np
```

# Load Dataset

## Objective

Load the employee attrition dataset for data cleaning.

```python
df = pd.read_csv("../data/employee_attrition.csv")

df.head()
```

## Observation

The dataset was successfully loaded and is ready for cleaning.

# Create a Backup Copy

## Objective

Create a copy of the original dataset to preserve the raw data during cleaning.

```python
df_clean = df.copy()
```

## Observation

A backup copy of the dataset was created for cleaning operations.

# Check Dataset Shape

## Objective

Record the dataset dimensions before performing any cleaning operations.

```python
df_clean.shape
```

## Observation

The dataset shape was recorded and will be compared after cleaning.

# Remove Duplicate Rows

## Objective

Identify and remove duplicate records to improve data quality.

```python
# Check duplicate rows

df_clean.duplicated().sum()
```

```python
# Remove duplicates if any

df_clean.drop_duplicates(inplace=True)
```

```python
# Verify duplicate removal

df_clean.duplicated().sum()
```

## Observation

No duplicate records were found in the dataset.

# Handle Missing Values

## Objective

Check for missing values and handle them appropriately to ensure data completeness.

```python
# Check missing values

df_clean.isnull().sum()
```

## Observation

No missing values were found in the dataset. No imputation was required.

# Convert Data Types

## Objective

Verify that all columns have appropriate data types and convert them if necessary.

```python
# Check data types

df_clean.info()
```

## Observation

All columns have appropriate data types. No conversions were required.

# Clean Categorical Values

## Objective

Standardize categorical values by removing extra spaces and ensuring consistent capitalization.

```python
# Standardize Department values

df_clean["Department"] = (
    df_clean["Department"]
    .str.strip()
    .str.title()
)

# Verify cleaned values

df_clean["Department"].value_counts()
```

## Observation

Categorical values were standardized to ensure consistent formatting.

# Remove Constant Columns

## Objective

Remove columns that contain only one unique value, as they do not contribute useful information for analysis or modeling.

```python
# Check unique values

df_clean.nunique()
```

```python
# Remove constant columns

df_clean.drop(
    columns=[
        "EmployeeCount",
        "Over18",
        "StandardHours"
    ],
    inplace=True
)
```

```python
# Verify remaining columns

df_clean.columns
```

## Observation

Constant columns were removed as they do not provide any predictive value.

# Check Invalid Values

## Objective

Verify that numerical features contain realistic values and identify any invalid entries.

```python
# Summary statistics for key numerical columns

df_clean[["Age", "MonthlyIncome", "TotalWorkingYears"]].describe()
```

## Observation

The numerical values appear realistic. No invalid values were identified.

# Outlier Review

## Objective

Determine whether detected outliers should be removed or retained.

```python
# Review summary statistics for numerical columns

df_clean.describe().T
```

## Observation

Outliers were retained because they represent genuine employee differences rather than data errors.

# Verify Cleaned Dataset

## Objective

Verify the dataset after cleaning to ensure all quality checks have been addressed.

```python
df_clean.info()
```

```python
df_clean.describe().T
```

```python
df_clean.isnull().sum()
```

```python
df_clean.duplicated().sum()
```

## Observation

The dataset is clean with no missing values, duplicate records, or data type issues.

# Compare Before and After Cleaning

## Objective

Compare the dataset before and after the cleaning process.

```python
comparison = pd.DataFrame({
    "Check": [
        "Rows",
        "Missing Values",
        "Duplicate Rows",
        "Constant Columns",
        "Invalid Data Types"
    ],
    "Before": [
        df.shape[0],
        df.isnull().sum().sum(),
        df.duplicated().sum(),
        3,
        0
    ],
    "After": [
        df_clean.shape[0],
        df_clean.isnull().sum().sum(),
        df_clean.duplicated().sum(),
        0,
        0
    ]
})

comparison
```

## Observation

The comparison confirms that the dataset has been successfully cleaned while preserving all valid records.

# Save Clean Dataset

## Objective

Save the cleaned dataset as a new file without modifying the original dataset.

```python
df_clean.to_csv(
    "../data/employee_attrition_clean.csv",
    index=False
)
```

## Observation

The cleaned dataset was successfully saved for use in the next stage of the project.

# Data Cleaning Summary

Duplicate records were checked and removed where necessary.

Missing values were inspected, and no imputation was required.

Data types were verified and found to be appropriate.

Categorical values were standardized to ensure consistency.

Constant columns (**EmployeeCount**, **Over18**, and **StandardHours**) were removed because they provide no predictive value.

Outliers were retained because they represent valid employee characteristics rather than data errors.

The cleaned dataset was saved as **employee_attrition_clean.csv** for use in subsequent analysis.
